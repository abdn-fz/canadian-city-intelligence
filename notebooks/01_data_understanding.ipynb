### Data Understanding and Cleaning

This notebook focuses on understanding, inspecting, and preparing the datasets used in the Canadian City Intelligence project.

The project analyzes the evolution of major Canadian cities using indicators related to:

The objective of this project is to analyze how major Canadian cities have evolved over time across different socioeconomic indicators.

This project combines data analysis, SQL, machine learning and Power BI.

The project uses five CSV datasets covering major Canadian cities.

The datasets contain information about:

- Consumer Price Index (CPI)

## 3. Import Libraries

The following Python libraries are used for data loading, manipulation, and analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 4. Load the Datasets

The five CSV files are loaded into separate Pandas DataFrames.

The `latin1` encoding is used because some city names contain accented characters, such as Montréal.

```python
files = {
    "CPI": "../data/raw/CPI_CMA_2006_2024.csv",
    "Crime": "../data/raw/crime_rate_CMA_2006_2023.csv",
    "Housing": "../data/raw/housing_price_index_by_CMA_2006_2023.csv",
    "Population": "../data/raw/population_by_CMA_2006_2023.csv",
    "Unemployment": "../data/raw/unemployment_by_CMA_2006_2024.csv"
}

datasets = {}

for name, file in files.items():
    datasets[name] = pd.read_csv(
        file,
        encoding="latin1"
    )
```

## 5. Preview the Data

The first observations of each dataset are reviewed to understand how the information is structured and to identify potential formatting issues.

```python
for name, df in datasets.items():
    print(f"\n--- {name} ---")
    display(df.head())
```

## 6. Initial Dataset Inspection

Before cleaning the data, the structure and dimensions of each dataset are examined.

This helps identify the number of observations, variables, and the temporal structure of each dataset.

```python
for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```

The datasets contain information for 9 Canadian cities.

- CPI: 9 cities and 234 monthly observations per city.
- Crime: 9 cities with annual data from 2006 to 2023.
- Housing: 9 cities with monthly observations.
- Population: 9 cities with annual data from 2006 to 2024.
- Unemployment: 9 cities with annual data from 2006 to 2024.

The datasets cover multiple socioeconomic indicators with different time granularities. This difference in temporal granularity will be considered during the cleaning and transformation stage.

## 7. Data Type Inspection

The data types are inspected to identify potential issues before data transformation.

```python
for name, df in datasets.items():
    print(f"\n--- {name} ---")
    print("Missing values:", df.isna().sum().sum())
    print("Data types:")
    print(df.dtypes.value_counts())
```

The initial inspection shows that the datasets have different data type structures. CPI and Unemployment contain mostly numeric variables, while Crime and Population have all temporal columns stored as `object`. Housing also contains a mix of numeric and non-numeric column types.

## 8. Check City Names

Since the datasets will later be combined, it is important to verify that the city names are consistent across all files.

```python
for name, df in datasets.items():
    print(f"{name}:")
    print(df["City"].tolist())
    print()
```

The five datasets contain the same 9 Canadian cities, which will allow them to be combined during the analysis.

One naming inconsistency was identified in the Housing dataset: `Montréal, Quebec` is used instead of `Montréal`.

This inconsistency will be standardized during the data cleaning and transformation stage to ensure consistent city identifiers across all datasets.

## 9. Missing Values

Missing values are examined in each dataset before deciding how they should be handled.

No values are removed or replaced without first understanding their meaning.

```python
for name, df in datasets.items():
    print(f"\n--- {name} ---")
    print(df.isna().sum().sort_values(ascending=False).head(10))
```

The initial missing-value check does not detect any `NaN` values in the raw datasets.

However, this result does not allow us to conclude that the datasets contain no missing values. Some variables are not stored with consistent data types, and values that should be numerical may be stored as text. In these cases, missing values may be represented by other forms, such as empty strings or text placeholders, rather than recognized as `NaN` by Pandas.

Therefore, the absence of detected `NaN` values should be interpreted with caution. Missing-value validation will be revisited after the data types and formats have been standardized during the data cleaning and transformation stage.

The data type inconsistencies will also be investigated and corrected during the cleaning stage.

## 10. Duplicate Records

The datasets are checked for duplicated rows.

Duplicate records could affect aggregations, visualizations, and machine learning results.

```python
for name, df in datasets.items():
    print(f"{name}: {df.duplicated().sum()} duplicated rows")
```

No duplicate rows were found in any of the five datasets.

This indicates that each row represents a unique record within its respective dataset. No duplicate removal is required at this stage.

## 11. Temporal Granularity

The datasets do not all have the same temporal granularity:

```python
for name, df in datasets.items():
    print(f"\n--- {name} ---")
    print("First temporal column:", df.columns[1])
    print("Last temporal column:", df.columns[-1])
    print("Number of temporal columns:", len(df.columns) - 1)
```

The datasets cover different time periods and have different temporal granularities:

- CPI: Monthly data from January 2006 to June 2025.
- Crime: Annual data from 2006 to 2023.
- Housing: Monthly data from January 2006 to May 2025.
- Population: Annual data from 2006 to 2024.
- Unemployment: Annual data from 2006 to 2024.

So the datasets can be divided into two temporal granularities:

- Monthly datasets: CPI and Housing.
- Annual datasets: Crime, Population, and Unemployment.

This distinction is important because the monthly datasets contain more detailed time-series information, while the annual datasets provide one observation per year.

The datasets will therefore be kept at their original granularity during the initial cleaning stage. Aggregation to a common yearly level will only be performed when required.

Based on the initial inspection, the following transformations will be performed:

1. Standardize column names where necessary.
2. Convert relevant columns to appropriate numeric data types.
3. Reshape annual datasets from wide format to long format.
4. Reshape monthly datasets into a structured date format.
5. Verify and standardize city names.
6. Handle missing values according to their meaning.
7. Create clean analytical tables for further analysis.

The monthly and annual datasets will initially remain separate because they have different temporal granularities.
